## 2.1 Dasar Teori
Python memiliki berbagai jenis operator yang digunakan untuk melakukan proses tertentu terhadap data dan variabel. Operator merupakan simbol atau kata khusus yang digunakan untuk melakukan operasi, seperti perhitungan, perbandingan, pemberian nilai, maupun pengolahan kondisi. Penggunaan operator menjadi salah satu dasar penting dalam pemrograman karena hampir setiap program membutuhkan operasi terhadap data.


2.1 Operator Aritmatika
Operator aritmatika berfungsi untuk mengolah nilai numerik dalam suatu perhitungan. Python mendukung operasi penjumlahan (+), pengurangan (-), perkalian (*), pembagian biasa (/), pembagian yang hasilnya dibulatkan ke bawah (//), sisa hasil bagi (%), serta pemangkatan (**). Urutan pengerjaannya mengikuti aturan prioritas matematika pada umumnya.

 2.2 Operator Perbandingan
Operator ini dipakai ketika dua nilai perlu dibandingkan satu sama lain, dan hasilnya selalu berupa nilai boolean, yaitu True atau False. Termasuk di dalamnya adalah operator kesetaraan (==, !=) dan operator relasional (>, <, >=, <=). Operator jenis ini banyak dipakai dalam pembuatan struktur percabangan seperti if-else.

 2.3 Operator Logika
Ketika sebuah program perlu mengecek lebih dari satu syarat sekaligus, operator logika berperan penting. and menghasilkan True hanya jika seluruh kondisi terpenuhi, or cukup satu kondisi saja yang benar, sedangkan not berfungsi membalikkan nilai kebenaran suatu ekspresi.

 2.4 Operator Penugasan
Operator penugasan digunakan untuk menyimpan atau memperbarui nilai suatu variabel. Selain tanda sama dengan (=) sebagai bentuk dasar, Python juga menyediakan operator gabungan seperti +=, -=, *=, dan /= yang memungkinkan penulisan kode menjadi lebih ringkas.

 2.5 Operator Bitwise
Berbeda dari operator lain yang bekerja pada nilai secara utuh, operator bitwise memproses data pada tingkat bit biner. Operator ini mencakup AND (&), OR (|), XOR (^), NOT (~), serta pergeseran bit ke kiri (<<) dan ke kanan (>>). Operator bitwise umumnya dipakai dalam pemrograman tingkat rendah atau optimisasi performa.

 2.6 Operator Identitas dan Keanggotaan
Operator identitas, yaitu is dan is not, digunakan untuk mengetahui apakah dua variabel menunjuk pada objek yang persis sama di dalam memori, bukan sekadar memiliki nilai yang sama. Sementara itu, operator keanggotaan in dan not in dipakai untuk memeriksa apakah suatu elemen terdapat di dalam struktur data seperti list, tuple, atau string.

````python
# Buat file dengan nama assignment_2611533028
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_3028
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_3028 = int(input("Input angka-1: "))
angka2_3028 = int(input("Input angka-2: "))

print("\nNilai awal angka1_3028 =", angka1_3028 )
print("Nilai angka2_3028 =", angka2_3028 )

# Assignment biasa
hasil_3028 = angka1_3028
print("\nAssignment biasa (=)")
print("Hasil =", hasil_3028)

# Assignment penambahan 
hasil_3028 = angka1_3028
hasil_3028 += angka2_3028
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_3028)

# Assignment pengurangan
hasil_3028 = angka1_3028
hasil_3028 -= angka2_3028
print("\nAssignment penambahan (-=)")
print("Hasil =", hasil_3028)

# Assignment perkalian
hasil_3028 = angka1_3028
hasil_3028 *= angka2_3028
print("\nAssignment penambahan (*=)")
print("Hasil =", hasil_3028)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3028 != 0:
    hasil_3028 = angka1_3028
    hasil_3028 /= angka2_3028
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_3028)
    # Operator tambahan
    hasil_3028 = angka1_3028
    hasil_3028 //= angka2_3028
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_3028)
    hasil_3028 = angka1_3028
    hasil_3028 %= angka2_3028
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_3028)
else:
    print("\nPembagian tidak dapat dilakukan")
    print("Angka kedua tidak boleh bernilai 0. ")

# Operator tambahan: assignment perpangkatan
hasil_3028 = angka1_3028
hasil_3028 **= angka2_3028
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_3028)
````

Mengecek apakah total harga minimal Rp150.000. Hasilnya berupa True atau False.

6. Mengecek jumlah tiket
`syarat_jumlah_3028 = jumlah_tiket_3028 >= 2`
Mengecek apakah penonton membeli minimal 2 tiket.

7. Mengecek status member
`status_valid_3028 = status_member_3028 == "member"`
Mengecek apakah status keanggotaan sama dengan "member". Jika sama persis, hasilnya True.

8. Mengecek voucher
`voucher_valid_3028 = kode_voucher_3028 in daftar_voucher_3028`
Operator `in` digunakan untuk mengecek apakah kode voucher yang dimasukkan terdapat di dalam list `daftar_voucher_3028`.

9. Menampilkan hasil validasi
```
print(f"Harga >= Rp150000       : {syarat_harga_3028}")
print(f"Jumlah Tiket >= 2       : {syarat_jumlah_3028}")
print(f"Status Member           : {status_valid_3028}")
print(f"Voucher Tersedia        : {voucher_valid_3028}")
print(f"Mendapatkan Diskon      : {syarat_harga_3028 or syarat_jumlah_3028}")
print(f"Mendapatkan Voucher     : {voucher_valid_3028}")
```
Menampilkan hasil dari semua pengecekan. `or` digunakan agar penonton dapat diskon jika **salah satu** syarat harga atau jumlah tiket terpenuhi.

10. Menghitung diskon dan total pembayaran
```
diskon_3028 = 0.10
print(f"Diskon                  : Rp{diskon_3028 * total_harga_3028}")
print(f"Total Pembayaran        : Rp{total_harga_3028 - total_harga_3028 * diskon_3028}")
print(f"Rata-rata Harga Tiket   : Rp{total_harga_3028 / jumlah_tiket_3028}")
```
Diskon ditetapkan 10% dari total harga. Total pembayaran adalah harga dikurangi diskon, dan rata-rata harga tiket dihitung dengan pembagian.

11. Menampilkan info hak akses
```
print("Kode Hak Akses          : bit0=Member, bit1=Harga, bit2=Jumlah, bit3=Voucher")
print(f"Member Access           : {status_valid_3028}")
print(f"Voucher Access          : {voucher_valid_3028}")
```
Menjelaskan bahwa tiap syarat akan diwakili oleh satu bit (posisi) dalam kode biner.

12. Membentuk kode biner 4-bit
```
kode_reservasi_3028 = int(status_valid_3028) << 0 | int(syarat_harga_3028) << 1 | int(syarat_jumlah_3028) << 2 | int(voucher_valid_3028) << 3
kode_referensi_3028 = int(status_valid_3028) << 0 | int(syarat_harga_3028) << 1 | int(voucher_valid_3028) << 3
```
Setiap status boolean diubah jadi 0/1 (`int(...)`), lalu digeser (`<<`) ke posisi bit-nya masing-masing, kemudian digabung dengan `|` (OR) menjadi satu angka biner 4-bit. `kode_referensi_3028` sengaja tidak menyertakan bit jumlah tiket (bit2), untuk dibandingkan nanti.

13. Menampilkan rincian dan hasil kode biner
```
print(f"{format(int(status_valid_3028) << 0, '04b')} | {format(int(syarat_harga_3028) << 1, '04b')} | ...")
print(f"Kode Biner              : {format(kode_reservasi_3028, '04b')}")
print(f"Kode Desimal            : {kode_reservasi_3028}")
```
`format(x, '04b')` menampilkan angka dalam bentuk biner 4 digit. Baris ini memperlihatkan tiap bit sebelum digabung, lalu hasil gabungannya.

14. Operasi AND — cek status member
```
print(f"Hasil Biner    : {format(kode_reservasi_3028 & (int(status_valid_3028) << 0), '04b')}")
```
Operator `&` (AND) dipakai untuk mengecek apakah bit member (bit0) aktif di `kode_reservasi_3028`. Jika hasilnya bukan 0, berarti status member aktif.

15. Operasi AND — cek voucher
```
print(f"Hasil Biner    : {format(kode_reservasi_3028 & (int(voucher_valid_3028) << 3), '04b')}")
```
Sama seperti sebelumnya, tapi memeriksa bit voucher (bit3).

16. Operasi XOR — bandingkan dua kode
```
print(f"Hasil Biner    : {format(kode_reservasi_3028 ^ kode_referensi_3028, '04b')}")
```
Operator `^` (XOR) membandingkan `kode_reservasi_3028` dan `kode_referensi_3028`. Bit yang berbeda antara keduanya akan bernilai 1, sedangkan bit yang sama bernilai 0 — di sini akan terlihat perbedaan pada bit jumlah tiket (bit2), karena `kode_referensi_3028` tidak menyertakannya.

17. Operasi Shift — upgrade level akses
```
print(f"Hasil Biner    : {format(kode_reservasi_3028 << 1, '04b')}")
```
Operator `<<` menggeser semua bit `kode_reservasi_3028` satu posisi ke kiri, mensimulasikan "kenaikan level akses" (nilainya jadi dua kali lipat).

18. Penutup
`print("\n=== SELESAI ===")`
Menandakan program telah selesai dijalankan.
