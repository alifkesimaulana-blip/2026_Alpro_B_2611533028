## 2.1 Dasar Teori

1. Perulangan for. Perulangan for mengeksekusi blok kode untuk setiap elemen dalam suatu urutan. Bentuk umumnya for variabel in urutan:, dan urutan angka biasanya dibuat dengan range().

2. Fungsi range(). Fungsi ini memiliki tiga bentuk:
<br>
a. range(stop): dari 0 sampai stop - 1.
<br>
b. range(start, stop): dari start sampai stop - 1.
<br>
c. range(start, stop, step): dengan lompatan step. Nilai step negatif membuat hitungan menurun, misalnya range(4, 0, -1) menghasilkan 4, 3, 2, 1.

3. Perulangan bersarang. Loop luar biasanya mengatur baris, sedangkan loop dalam mengatur isi tiap baris (kolom). Loop dalam dijalankan penuh pada setiap satu putaran loop luar.

4. Parameter end pada print(). Secara bawaan print() pindah baris setelah mencetak. Dengan end="" atau end=" ", hasil cetak tetap berada di baris yang sama. print() kosong dipakai untuk pindah baris.

5. Rumus pada pola jam pasir (N adalah ukuran, baris adalah nomor baris):
<br>
Jumlah karakter = pada bingkai: 4N + 5.
<br>
Spasi penyeimbang kiri dan kanan tiap baris: 2 × (N − baris).
<br>
Spasi kiri dan kanan pada titik pusat: 2N + 1.
<br>
Jumlah seluruh baris isi: 2N + 1, terdiri dari N baris atas, 1 baris pusat, dan N baris bawah.
<br>

6. Palindromik. Deretan angka yang dibaca dari kiri maupun kanan memiliki urutan yang sama terhadap poros <*>, misalnya 4 3 2 1 <*> 1 2 3

Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).
Struktur Pembentukan Pola (3 Fase + 2 Bingkai Horizontal):
<br>
<br>
<br>
 Bingkai Pembatas Horizontal (Border Atas dan Bawah):
<br>
○         Karakter pagar sudut (#) di ujung kiri dan kanan.
<br>
○         Karakter garis ganda (=) di bagian tengah sebanyak (4 * N + 5) karakter.
<br>
 Fase 1: Jam Pasir Atas (Reduksi Angka Menurun: Baris N turun s.d. 1):
<br>
○         Sisi kiri dibatasi garis tegak (|) dan satu spasi padding.
<br>
○         Spasi penyeimbang kiri sebanyak 2 * (N - baris).
<br>
○         Deret angka mundur dari baris turun ke 1, dipisahkan oleh spasi (contoh baris 4: "4 3 2 1 ").
<br>
○         Poros kristal tengah: karakter "<*>".
<br>
○         Deret angka maju dari 1 naik ke baris, diawali spasi (contoh baris 4: " 1 2 3 4").
<br>
○         Spasi penyeimbang kanan sebanyak 2 * (N - baris).
<br>
○         Satu spasi padding dan garis tegak (|) di sisi kanan.
<br>
 Fase 2: Poros Titik Pusat Jam Pasir (Titik Nol / Singularity):
<br>
○         Karakter garis tegak pembatas (|) di kiri.
<br>
○         Spasi penyeimbang kiri sebanyak (2 * N + 1).
<br>
○         Karakter poros kristal tunggal "<*>".
<br>
○         Spasi penyeimbang kanan sebanyak (2 * N + 1).
<br>
○         Karakter garis tegak pembatas (|) di kanan.
<br>
 Fase 3: Jam Pasir Bawah (Ekspansi Angka Menaik: Baris 1 naik s.d. N):
<br>
○         Sisi kiri dibatasi garis tegak (|) dan satu spasi padding.
<br>
○         Spasi penyeimbang kiri sebanyak 2 * (N - baris).
<br>
○         Deret angka mundur dari baris turun ke 1, dipisahkan oleh spasi.
<br>
○         Poros kristal tengah: karakter "<*>".
<br>
○         Deret angka maju dari 1 naik ke baris, diawali spasi.
<br>
○         Spasi penyeimbang kanan sebanyak 2 * (N - baris).
<br>
○         Satu spasi padding dan garis tegak (|) di sisi kanan.

a. Kode Program Tugas5_2611533028

```python
# tugas5_2611533028.py
# Pola Jam Pasir Kristal Palindromik Berbingkai (Pekan 5)
# Program dijalankan 2 kali berturut-turut (Uji Coba 1 dan Uji Coba 2)
# Menggunakan perulangan for bersarang murni (tanpa perkalian string)

for uji_3028 in range(2):
    print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
    n_3028 = int(input("Masukkan ukuran skala jam pasir (N): "))

    # ---------- Bingkai atas ----------
    print("#", end="")
    for garis_3028 in range(4 * n_3028 + 5):
        print("=", end="")
    print("#", end="")
    print()

    # ---------- Fase 1: jam pasir atas (baris N turun s.d. 1) ----------
    for baris_3028 in range(n_3028, 0, -1):
        print("| ", end="")
        for spasi_3028 in range(2 * (n_3028 - baris_3028)):
            print(" ", end="")
        for angka_3028 in range(baris_3028, 0, -1):
            print(angka_3028, end=" ")
        print("<*>", end="")
        for angka_3028 in range(1, baris_3028 + 1):
            print(" ", end="")
            print(angka_3028, end="")
        for spasi_3028 in range(2 * (n_3028 - baris_3028)):
            print(" ", end="")
        print(" |", end="")
        print()

    # ---------- Fase 2: poros titik pusat ----------
    print("|", end="")
    for spasi_3028 in range(2 * n_3028 + 1):
        print(" ", end="")
    print("<*>", end="")
    for spasi_3028 in range(2 * n_3028 + 1):
        print(" ", end="")
    print("|", end="")
    print()

    # ---------- Fase 3: jam pasir bawah (baris 1 naik s.d. N) ----------
    for baris_3028 in range(1, n_3028 + 1):
        print("| ", end="")
        for spasi_3028 in range(2 * (n_3028 - baris_3028)):
            print(" ", end="")
        for angka_3028 in range(baris_3028, 0, -1):
            print(angka_3028, end=" ")
        print("<*>", end="")
        for angka_3028 in range(1, baris_3028 + 1):
            print(" ", end="")
            print(angka_3028, end="")
        for spasi_3028 in range(2 * (n_3028 - baris_3028)):
            print(" ", end="")
        print(" |", end="")
        print()

    # ---------- Bingkai bawah ----------
    print("#", end="")
    for garis_3028 in range(4 * n_3028 + 5):
        print("=", end="")
    print("#", end="")
    print()
```

Penjelasan: 
1. for uji_3028 in range(2): seluruh program diulang 2 kali (Uji Coba 1 dan 2). Tiap putaran meminta input N baru.
2. n_3028 = int(input(...)): meminta ukuran N dari pengguna, lalu diubah jadi integer.
3. Cetak #, lalu loop range(4 * n + 5) mencetak = satu per satu, lalu cetak # penutup. Untuk N=4 jumlahnya 21 karakter =.
4. print() kosong pindah ke baris berikutnya. Bingkai bawah memakai kode yang sama.
5. range(n, 0, -1): baris turun dari N ke 1.
6. Cetak | (garis tegak dan satu spasi).
7. Loop spasi sebanyak 2 * (n - baris) membuat baris makin menjorok ke dalam.
8. range(baris, 0, -1) mencetak angka mundur. Contoh baris 4: 4 3 2 1 .
9. Cetak poros <*>.
10. range(1, baris + 1) mencetak angka maju, tiap angka didahului spasi. Contoh: 1 2 3 4.
11. Loop spasi kanan (jumlahnya sama dengan sisi kiri), lalu cetak | dan print().
