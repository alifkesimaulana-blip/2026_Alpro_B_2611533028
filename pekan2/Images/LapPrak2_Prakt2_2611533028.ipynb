## 2.1 Dasar Teori

1. Algoritma dan Pemrograman

Algoritma merupakan serangkaian langkah-langkah yang disusun secara logis, terstruktur, dan sistematis untuk menyelesaikan suatu permasalahan. Algoritma menjadi dasar dalam pembuatan sebuah program karena menentukan bagaimana suatu masalah dapat diselesaikan melalui langkah-langkah tertentu. Pemrograman merupakan proses menerapkan algoritma ke dalam bahasa yang dapat dipahami dan dijalankan oleh komputer. Melalui pemrograman, seseorang dapat memberikan instruksi kepada komputer untuk melakukan tugas tertentu sesuai dengan tujuan yang telah ditentukan.

2. Bahasa Pemrograman Python

Python merupakan salah satu bahasa pemrograman tingkat tinggi yang banyak digunakan dalam berbagai bidang. Python dikenal sebagai bahasa pemrograman yang memiliki sintaks sederhana, mudah dibaca, dan mudah dipahami, sehingga cocok digunakan untuk mempelajari dasar-dasar pemrograman. Python dapat digunakan dalam berbagai bidang, seperti pengembangan aplikasi, website, pengolahan data, kecerdasan buatan, otomatisasi, dan berbagai kebutuhan teknologi lainnya. Kemudahan dalam penggunaan serta perkembangan yang luas menjadikan Python sebagai salah satu bahasa pemrograman yang banyak dipelajari.

3. Variabel

Variabel merupakan nama atau tempat yang digunakan untuk menyimpan suatu nilai atau data di dalam program. Data yang disimpan dalam variabel dapat digunakan kembali dan diolah sesuai dengan kebutuhan program. Dalam bahasa pemrograman Python, variabel digunakan untuk mempermudah proses penyimpanan dan pengolahan data. Nilai yang terdapat dalam sebuah variabel dapat berubah sesuai dengan instruksi yang diberikan dalam program.

4. Tipe Data

Tipe data merupakan jenis atau kategori dari nilai yang disimpan dalam sebuah variabel. Setiap tipe data memiliki karakteristik yang berbeda dan digunakan sesuai dengan jenis informasi yang akan diolah. Dalam bahasa Python, terdapat beberapa tipe data dasar yang sering digunakan, yaitu Integer, Float, String, dan Boolean. Integer digunakan untuk bilangan bulat, Float digunakan untuk bilangan desimal, String digunakan untuk data berupa teks atau karakter, sedangkan Boolean digunakan untuk nilai logika yang memiliki dua kemungkinan, yaitu benar atau salah.
Pemahaman mengenai tipe data sangat penting karena jenis data yang digunakan akan memengaruhi proses pengolahan data dalam sebuah program.


5. Hubungan Variabel dan Tipe Data

Variabel dan tipe data merupakan dua konsep dasar yang saling berkaitan dalam pemrograman. Variabel berfungsi sebagai tempat untuk menyimpan informasi, sedangkan tipe data menunjukkan jenis dari informasi atau nilai yang disimpan tersebut. Dengan memahami penggunaan variabel dan tipe data, seorang programmer dapat mengelola data dengan lebih baik serta menentukan proses yang tepat dalam pembuatan sebuah program.



6. Pentingnya Memahami Variabel dan Tipe Data

Pemahaman mengenai variabel dan tipe data merupakan salah satu dasar penting dalam mempelajari bahasa pemrograman. ### 2.2.1 Program Konstanta_2611533028

```python
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_3028

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3028 = float(input('Masukkan nilai jari-jari: '))
luas_3028 = PI * jari_3028 * jari_3028
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3028, luas_3028))
```

**Penjelasan :**
1. PI: Final = 3.14 — bikin konstanta PI (nilainya 3.14, tidak boleh diubah lagi, ditandai dengan Final).
2. print("pi: %f" % (PI)) — nampilin nilai PI.
3. jari_3028 = float(input(...)) — minta user masukkan jari-jari lingkaran, diubah jadi angka desimal.
4. luas_3028 = PI * jari_3028 * jari_3028 — rumus luas lingkaran (π × r × r).
5. print(...) — tampilkan hasil luasnya, dibulatkan 2 angka di belakang koma.

### 2.2.1 Program Numerik_2611533028

```python
# nama variabel ditambah 4 digit terakhir NIM contoh:a_3028
# create a variabel with numeric type value
a_3028=100
print("The type of variabel having value", a_3028," is ", type(a_3028))

#create a variabel with float value
b_3028=10.2345
print("The type of variabel having value", b_3028," is ", type(b_3028))

#create a variabel with complex value
c_3028=100+3j
print("The type of having value", c_3028, " is ", type(c_3028))
```

**Penjelasan :**
1. Baris 1–2 — membuat variabel a_3028 bertipe integer (bilangan bulat), nilainya 100. Lalu type() dipakai untuk menampilkan jenis tipe datanya.
2. Baris 3–4 — membuat variabel b_3028 bertipe float (bilangan desimal), nilainya 10.2345, lalu ditampilkan tipenya.
3. Baris 5–6 — membuat variabel c_3028 bertipe complex (bilangan kompleks, ada bagian imajiner 3j), lalu ditampilkan tipenya.
