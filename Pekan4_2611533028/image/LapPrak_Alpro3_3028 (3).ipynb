## 2.1 Dasar Teori

2.1 Pernyataan If
Pernyataan if merupakan bentuk paling dasar dari struktur percabangan. Blok kode di bawah if hanya akan dieksekusi apabila ekspresi kondisional yang diuji bernilai True. Apabila kondisi bernilai False, seluruh baris pada blok tersebut dilewati tanpa ada tindakan apa pun.

2..2 Pernyataan If-Else
Struktur if-else menambahkan jalur alternatif yang akan dijalankan apabila kondisi pada if bernilai False. Dengan demikian, program akan selalu mengeksekusi salah satu dari dua blok yang tersedia, sehingga selalu ada keluaran yang dihasilkan.

2.3 Pernyataan If-Elif-Else Berjenjang (Nested)
Struktur if-elif-else digunakan ketika terdapat lebih dari dua kemungkinan hasil yang saling eksklusif. Python akan mengevaluasi setiap kondisi secara berurutan dari atas ke bawah, dan begitu satu kondisi terpenuhi, blok tersebut dijalankan lalu seluruh proses pengecekan dihentikan sehingga kondisi-kondisi berikutnya tidak lagi diperiksa.

2.4 Multi-If (Blok If Independen)
Berbeda dengan elif, penulisan beberapa blok if secara terpisah menyebabkan Python mengevaluasi setiap blok tersebut secara independen, tanpa saling menghentikan proses pengecekan satu sama lain. Apabila kondisi-kondisi yang diuji saling eksklusif, hasil akhirnya hanya satu blok yang aktif. Namun apabila kondisi-kondisi tersebut dapat bernilai True secara bersamaan, maka lebih dari satu blok dapat dijalankan sekaligus, seperti pada kasus akumulasi diskon belanja.

2.5 Pernyataan Match Case
Pernyataan match, yang diperkenalkan pada Python 3.10 melalui PEP 634, memungkinkan pencocokan sebuah nilai terhadap beberapa pola (pattern) secara berurutan menggunakan blok case. Struktur ini menjadi alternatif yang lebih terstruktur dibandingkan rangkaian if-elif yang panjang, khususnya ketika terdapat banyak nilai tetap (fixed values) yang perlu dibandingkan. Blok case _ berfungsi sebagai pola wildcard yang menangani nilai apa pun yang tidak cocok dengan pola-pola sebelumnya.

2.6 Fungsi Input dan Konversi Tipe Data
Seluruh program pada praktikum ini menggunakan fungsi input() untuk menerima data dari pengguna, yang secara default menghasilkan tipe data string. Oleh karena itu, diperlukan fungsi konversi seperti int() untuk bilangan bulat dan float() untuk bilangan desimal agar nilai tersebut dapat dibandingkan secara numerik di dalam ekspresi kondisional. Selain itu, metode string seperti .strip(), .lower(), dan pengindeksan [0] digunakan untuk menormalkan variasi masukan pengguna, misalnya perbedaan huruf besar-kecil atau spasi berlebih, sebelum dibandingkan dalam kondisi.

### 2.2.3 Program match_case1_2611533028

```python
# Buat file dengan nama match_case1_2611533028.py
# Buat program untuk match case 
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_3028 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_3028:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```


Penjelasan:
1. Input diubah jadi bilangan bulat (int) karena bulan berupa angka utuh (1, 2, 3, ...).
2. match mengambil nilai bulan_3028, lalu Python mencocokkannya dari atas ke bawah dengan tiap case.
Begitu ditemukan case yang nilainya sama persis dengan bulan_3028, blok di bawah case itu dijalankan, dan pencocokan berhenti (tidak lanjut cek case berikutnya).
3. Contoh: kalau bulan_3028 = 3, Python cek case 1 (tidak cocok) → case 2 (tidak cocok) → case 3 (cocok!) → cetak "Maret" → selesai.
4. case _ adalah kasus default/wildcard — artinya "kalau tidak cocok dengan semua case di atas". Jadi kalau pengguna input misalnya 13 atau 0, akan masuk ke sini dan cetak "Angka tidak valid".

### 2.2.5 Program multi_if2_2611533028

```python
# Buat file dengan nama multi_if2_2611533028.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3028 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y atau 'ya')
input_member_3028 = input("Apakah Anda Member? (y/t): ").strip().lower()
is_member_3028 = input_member_3028 in ["y", "ya"]

# Input satus kode promo (mengecek apakah user mengetik 'y atau 'ya')
input_promo_3028 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3028 = input_promo_3028 in ["y", "ya"]

total_diskon_persen_3028 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3028 > 1000000:
    total_diskon_persen_3028 += 10  # Diskon belanja besar 

if is_member_3028:
    total_diskon_persen_3028 += 5  # Diskon member

if kode_promo_valid_3028:
    total_diskon_persen_3028 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3028 = total_belanja_3028 * (total_diskon_persen_3028 / 100)
total_bayar_3028 = total_belanja_3028 - nominal_diskon_3028

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_3028}% (Rp {nominal_diskon_3028:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_3028:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_3028}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

Penjelasan:
1. input total belanja, jadi angka desimal.
2. .strip() menghapus spasi kosong di depan/belakang teks (misal kalau user tidak sengaja ketik " y ").
3. .lower() mengubah semua huruf jadi huruf kecil (misal "Y" atau "YA" jadi "y"/"ya"), supaya tidak peduli besar-kecil huruf.
4. in ["y", "ya"] mengecek apakah teks yang diketik user ada di dalam daftar ["y", "ya"]. Hasilnya berupa True atau False, disimpan di is_member_3028.
5. Sama seperti di atas, tapi untuk mengecek kode promo.
6. Variabel penampung total diskon, dimulai dari 0%.
7. ni bagian kunci multi-if: tiga if terpisah, masing-masing dicek sendiri-sendiri, tidak saling menghalangi.
+= berarti "tambahkan ke nilai yang sudah ada". Jadi kalau ketiga syarat terpenuhi, diskon akan bertambah/menumpuk: 10 + 5 + 15 = 30%.
8. Ini beda dengan program nomor 4 tadi — di sini diskon memang dirancang untuk bisa akumulasi, sedangkan di nomor 4 empat syarat itu saling eksklusif jadi cuma satu yang jalan.
9. Menghitung nilai rupiah dari diskon: persen dibagi 100 dulu jadi desimal (misal 30% jadi 0.3), lalu dikali total belanja.
10. Total bayar = belanja awal dikurangi nominal diskon.
11. f adalah f-string, cara memasukkan nilai variabel langsung ke dalam teks pakai {}.
12. 0f adalah format angka: , menambahkan pemisah ribuan (misal 1,000,000), .0f artinya tampilkan tanpa angka desimal (dibulatkan).

### 2.2.6 Program nilai_2611533028

```python
# Buat file dengan nama nilai_2611533028.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai_3028 = int(input("Inputkan nilai angka= "))

if nilai_3028 >= 81:
    print("A")
elif nilai_3028 >= 70:
    print("B")
elif nilai_3028 >= 60:
    print("C")
elif nilai_3028 >= 50:
    print("D")
else:
    print("E")
```

Penjelasan:

4. Kalau tidak (misal nilainya 75), Python lanjut cek elif nilai_3028 >= 70 → benar → cetak "B" → berhenti.
5. Begitu seterusnya sampai ke else di paling bawah, yang jadi "jaring pengaman" kalau semua syarat di atas gagal (nilai di bawah 50).

```python
# liburan_keluar_kota_2611533028.py
# Program Sistem Reservasi Liburan Keluar Kota
# Menggabungkan semua jenis conditional Python dalam satu program:
# - match case          -> memilih paket destinasi
# - if-elif bersarang    -> mengecek kelayakan sewa kendaraan di lokasi (di dalam tiap case)
# - multi-if             -> akumulasi diskon (bisa menumpuk)
# - if biasa             -> validasi input & bonus oleh-oleh

print("=== SISTEM RESERVASI LIBURAN KELUAR KOTA ===")

nama_3028 = input("Masukkan Nama Pemesan            : ")
umur_3028 = int(input("Input umur anda                  : "))
sim_c_3028 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].strip().lower()

print("\nPilihan Paket Destinasi (1-5):")
print("  1. Bali                 (Rp 3,000,000)")
print("  2. Bandung              (Rp 1,200,000)")
print("  3. Yogyakarta           (Rp 1,500,000)")
print("  4. Raja Ampat           (Rp 7,000,000)")
print("  5. Lombok All-Access    (Rp 2,500,000)")
destinasi_3028 = int(input("Masukkan nomor paket (1-5)       : "))
jumlah_orang_3028 = int(input("Masukkan jumlah orang            : "))

if jumlah_orang_3028 <= 0:
    print("Jumlah orang tidak valid!")

member_3028 = input("Apakah Anda member? (y/t)        : ")[0].strip().lower()
promo_3028 = input("Apakah kode promo valid? (y/t)   : ")[0].strip().lower()

harga_paket_3028 = 0
nama_destinasi_3028 = ""

print("\n--- KELAYAKAN SEWA KENDARAAN DI LOKASI ---")

# --- MATCH CASE: memilih destinasi, dengan pengecekan kelayakan (nested if) di tiap case ---
match destinasi_3028:
    case 1:  # Bali - sewa motor sendiri
        nama_destinasi_3028 = "Bali"
        if umur_3028 >= 17 and sim_c_3028 == "y":
            print("Anda sudah dewasa dan boleh menyewa motor sendiri di Bali.")
        elif umur_3028 >= 17 and sim_c_3028 != "y":
            print("Anda sudah dewasa tetapi tidak boleh sewa motor (wajib sewa mobil dengan supir).")
        elif umur_3028 < 17 and sim_c_3028 == "y":
            print("Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Anda belum cukup umur untuk menyewa motor sendiri.")
        harga_paket_3028 = 3000000

    case 2:  # Bandung - trekking Kawah Putih
        nama_destinasi_3028 = "Bandung"
        if umur_3028 >= 10:
            print("Anda boleh mengikuti trekking Kawah Putih sendiri.")
        else:
            print("Anda wajib didampingi orang tua.")
        harga_paket_3028 = 1200000

    case 3:  # Yogyakarta - sepeda gunung lereng Merapi
        nama_destinasi_3028 = "Yogyakarta"
        if umur_3028 >= 15:
            print("Anda boleh mengendarai sepeda gunung sendiri.")
        else:
            print("Anda wajib didampingi orang tua.")
        harga_paket_3028 = 1500000

    case 4:  # Raja Ampat - snorkeling dengan kapal
        nama_destinasi_3028 = "Raja Ampat"
        if umur_3028 >= 17:
            print("Anda sudah dewasa dan boleh ikut snorkeling tanpa pendamping.")
        else:
            print("Anda wajib didampingi orang tua saat naik kapal.")
        harga_paket_3028 = 7000000

    case 5:  # Lombok All-Access - motor trail + boat
        nama_destinasi_3028 = "Lombok All-Access"
        if umur_3028 >= 17 and sim_c_3028 == "y":
            print("Anda sudah dewasa dan boleh menyewa motor trail sendiri.")
        elif umur_3028 >= 17 and sim_c_3028 != "y":
            print("Anda sudah dewasa tetapi tidak boleh sewa motor trail (wajib didampingi instruktur).")
        elif umur_3028 < 17 and sim_c_3028 == "y":
            print("Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Anda belum cukup umur untuk menyewa motor trail sendiri.")
        harga_paket_3028 = 2500000

    case _:
        print("Paket destinasi tidak valid!")

# --- MULTI-IF: tiap syarat diskon dicek terpisah, jadi bisa menumpuk sekaligus ---
diskon_3028 = 0

if harga_paket_3028 * jumlah_orang_3028 >= 3000000:
    diskon_3028 += 10       # diskon rombongan belanja besar

if member_3028 in ["y", "ya"]:
    diskon_3028 += 5        # diskon member

if promo_3028 in ["y", "ya"]:
    diskon_3028 += 15       # diskon kode promo

if jumlah_orang_3028 >= 4:
    diskon_3028 += 5        # diskon rombongan minimal 4 orang

nominal_3028 = harga_paket_3028 * jumlah_orang_3028
nominal_diskon_3028 = nominal_3028 * (diskon_3028 / 100)
total_bayar_3028 = nominal_3028 - nominal_diskon_3028

print("\n--- Rincian Pembayaran ---")
print(f"Destinasi        : {nama_destinasi_3028}")
print(f"Subtotal Belanja : Rp {nominal_3028:,}")

# --- IF biasa: bonus oleh-oleh jika belanja di atas batas tertentu ---
if nominal_3028 > 5000000:
    print("Selamat! Anda berhak mendapatkan Paket Oleh-Oleh Gratis.")

print(f"Total Diskon     : {diskon_3028}% (Rp {nominal_diskon_3028:,})")
print(f"Total Bayar      : (Rp {total_bayar_3028:,})")
print("Catatan Layanan  : Terima kasih telah memesan liburan bersama kami.")
print("Program Selesai")
```

Penjelasan: 
1. print(...) di baris pertamanama_3028: disimpan sebagai teks/string, tidak perlu dikonversi.
2. umur_3028: dibungkus int(...) karena umur harus berupa angka bulat untuk bisa dibandingkan (>=, <) nanti.
3. sim_c_3028: ada tiga proses berantai di sini:
   input(...) mengambil jawaban user, misalnya "Ya" atau " y ".
   [0] mengambil karakter pertama saja dari teks tersebut (jadi "Ya" atau "yes" sama-sama jadi "Y").
   .strip() menghapus spasi kosong di depan/belakang.
   .lower() mengubah jadi huruf kecil, supaya "Y" dan "y" dianggap sama.
    Ketiganya dirangkai (chaining) agar hasil akhirnya konsisten, apa pun variasi ketikan user. hanya mencetak judul/header program sebagai tampilan awal, tidak memengaruhi logika apa pun.
4. \n di awal string artinya "buat baris kosong dulu" sebelum mencetak teks, supaya tampilannya tidak menempel dengan input sebelumnya.
5. destinasi_3028 dan jumlah_orang_3028 sama-sama diubah jadi int karena keduanya akan dipakai dalam operasi angka (pencocokan match dan perkalian harga).
6. ni if sederhana tanpa else. Kalau jumlah orang yang diinput 0 atau negatif (tidak masuk akal), program mencetak pesan error lalu exit() menghentikan program saat itu juga — baris-baris kode di bawahnya tidak akan dijalankan sama sekali.
7. Kalau jumlah orang valid (lebih dari 0), if ini dilewati dan program lanjut ke baris berikutnya seperti biasa.
8. Input member & promo disaring sama seperti sim_c_3028 sebelumnya — ambil huruf pertama, buang spasi, ubah ke huruf kecil — supaya "Y", "y", " Ya " semuanya dianggap sama.
9. match destinasi_3028 mencocokkan angka pilihan user dengan salah satu case. Begitu ketemu yang cocok, hanya blok itu yang dijalankan, sisanya diabaikan. Di dalam setiap case, ada dua hal yang terjadi:
10. if-elif-else (nested) mengecek apakah user boleh melakukan aktivitas di destinasi itu sendirian, berdasarkan umur (dan SIM C untuk Bali/Lombok yang butuh sewa motor). Ini cuma cetak pesan informasi, tidak menghentikan program.
11. harga_paket_3028 diisi sesuai destinasi yang dipilih.
12. Kalau angka yang diinput di luar 1–5, jatuh ke case _ → cetak error → menghentikan program.
13. Empat blok if terpisah (bukan elif) menghitung diskon. Karena masing-masing dicek sendiri-sendiri, nilainya bisa menumpuk lewat +=.
14.  Contoh: kalau belanja besar + member + promo valid + rombongan ≥4 orang, semuanya benar sekaligus → diskon jadi 10+5+15+5 = 35%.
15. Perhitungan akhir: nominal_3028 (harga × jumlah orang) dikurangi nominal_diskon_3028 (nominal dikali persen diskon dibagi 100) menghasilkan total_bayar_3028 — angka yang benar-benar harus dibayar.
16. f"..." (f-string) memasukkan nilai variabel langsung ke dalam teks lewat {}.
17. {nominal_3028:,} — tanda :, menambahkan pemisah ribuan otomatis (misalnya 3000000 tampil jadi "3,000,000").
18. if nominal_3028 > 5000000 adalah if tunggal tanpa else — kalau belanja di atas 5 juta, tampil bonus; kalau tidak, baris ini dilewati saja tanpa pesan apa pun (mirip if1_2611533028.py).
19. Dua baris terakhir selalu bercetak menandakan program sudah selesai.


<h1 align="center">BAB III<br>KESIMPULAN</h1>
Berdasarkan enam program yang telah dibuat, dapat disimpulkan bahwa Python menyediakan beragam bentuk struktur percabangan yang masing-masing memiliki karakteristik dan kegunaan berbeda. Pernyataan if tunggal cocok digunakan untuk pengujian satu syarat sederhana, sedangkan if-else digunakan ketika terdapat dua kemungkinan hasil yang saling meniadakan.
Struktur if-elif-else berjenjang tepat digunakan untuk kondisi yang saling eksklusif dan berurutan, seperti konversi nilai angka menjadi huruf, karena begitu satu syarat terpenuhi, pengecekan syarat berikutnya dihentikan. Sebaliknya, multi-if yang terdiri atas beberapa blok if terpisah memungkinkan setiap kondisi diperiksa secara independen, sehingga cocok digunakan pada kasus yang membutuhkan akumulasi, seperti penjumlahan persentase diskon belanja dari beberapa syarat yang terpenuhi sekaligus.
Sementara itu, pernyataan match case terbukti memberikan alternatif penulisan kode yang lebih ringkas dan mudah dibaca dibandingkan rangkaian if-elif yang panjang, khususnya untuk kasus pencocokan terhadap sejumlah nilai tetap seperti konversi angka bulan menjadi nama bulan. Dengan memahami perbedaan mendasar antar struktur ini, seorang pemrogram dapat memilih bentuk percabangan yang paling sesuai dengan logika persoalan yang dihadapi, sehingga program yang dihasilkan lebih efisien, akurat, dan mudah dipelihara.
